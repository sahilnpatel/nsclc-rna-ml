```python
# Import necessary libraries
from sklearn.preprocessing import StandardScaler
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import joblib # for saving the model

# Load your data
# Replace the file path with the path to the CSV file you just saved
data = pd.read_csv('data/source/legacy_ml.csv')

df = data 

# Drop columns with names like 'Unnamed: #' if they have no relevant data
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Drop rows where all columns are NaN
df = df.dropna(how='all')

# Optionally, you could also drop columns where all values are NaN
df = df.dropna(axis=1, how='all')

# Initialize scaler and standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df) # Scale all circRNA, miRNA, and mRNA features


# Define the number of clusters you want to find
n_clusters = 1000 # Adjust based on the patterns you're exploring


# Initialize and fit the K-means model
kmeans = KMeans(n_clusters=n_clusters, random_state=52)
kmeans.fit(X_scaled)


# Get cluster assignments for each sample
cluster_labels = kmeans.labels_


# Add cluster labels to the original data for analysis
data['Cluster'] = cluster_labels


# View the mean expression of each feature per cluster
cluster_means = data.groupby('Cluster').mean()
print(cluster_means)


# Reduce data to 2 dimensions for visualization with PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)


# Plot the clusters
plt.figure(figsize=(10, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', s=50)
plt.title("K-means Clustering of NSCLC Samples (PCA-reduced)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.colorbar(label='Cluster Label')
plt.show()


from sklearn.metrics import silhouette_score


silhouette_scores = []
wcss = []

for k in range(2, 50, 5): # Try a range of cluster numbers
    kmeans = KMeans(n_clusters=k, random_state=52)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))


# Plot Elbow Method
plt.figure(figsize=(8, 4))
plt.plot(range(2, 50, 5), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()


# Plot Silhouette Scores
plt.figure(figsize=(8, 4))
plt.plot(range(2, 50, 5), silhouette_scores, marker='o')
plt.title("Silhouette Scores")
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.show()
```
